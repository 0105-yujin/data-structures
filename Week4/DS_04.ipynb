## 0. 준비
이번 실습에서는 내부 표현을 숨기고, 정해진 메서드로만 자료구조를 사용합니다. 빈 구조에서 삭제·조회할 때는 `IndexError`를 발생시키겠습니다.

```python
from collections import deque

# 테스트 도우미: 오류가 예상되는 코드가 실제로 오류를 내는지 확인합니다.
def expect_error(error_type, func):
    try:
        func()
    except error_type:
        print(f"OK: {error_type.__name__} 발생")
    else:
        raise AssertionError(f"{error_type.__name__}가 발생해야 합니다.")
```

## 1. Stack ADT — LIFO
**Last-In, First-Out:** 가장 나중에 넣은 항목이 먼저 나온다.

구현할 인터페이스: `push`, `pop`, `peek`, `is_empty`, `size`

```python
class Stack:
    def __init__(self):
        # TODO 1: 빈 리스트를 내부 저장소로 만드세요.
        self.items = []

    def push(self, item):
        # TODO 2: item을 스택의 top에 추가하세요. (힌트: append)
        self.items.append(item)

    def pop(self):
        # TODO 3: 비어 있으면 IndexError를 발생시키고,
        # 그렇지 않으면 top 항목을 제거하여 반환하세요. (힌트: pop)
        if self.is_empty():
          raise IndexError("pop from empty stack")
        return self.items.pop()

    def peek(self):
        # TODO 4: 비어 있으면 IndexError를 발생시키고,
        # 그렇지 않으면 top 항목을 제거하지 않고 반환하세요.
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self.items[-1]

    def is_empty(self):
        # TODO 5: 비어 있는지 bool로 반환하세요.
        return len(self.items) == 0

    def size(self):
        # TODO 6: 원소 개수를 반환하세요.
        return len(self.items)
```

```python
# Stack 기본 테스트: TODO를 모두 구현한 뒤 실행하세요.
s = Stack()
assert s.is_empty() is True
s.push('A'); s.push('B'); s.push('C')
assert s.size() == 3
assert s.peek() == 'C'
assert s.pop() == 'C'
assert s.pop() == 'B'
assert s.pop() == 'A'
assert s.is_empty() is True
expect_error(IndexError, s.pop)
print('Stack 기본 테스트 통과!')
```

### 연습 1 — 괄호 검사
여는 괄호 `(`, `[`, `{`는 push하고, 닫는 괄호가 나오면 대응하는 여는 괄호가 top에 있는지 확인합니다.

- 닫는 괄호에서 스택이 비었거나 짝이 맞지 않으면 `False`
- 모든 문자를 처리한 뒤 스택이 비어 있으면 `True`

```python
def is_balanced(expression):
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = Stack()

    for ch in expression:
        # TODO 7: 여는 괄호를 처리하세요.
        if ch == '(' or ch == '[' or ch == '{':
            stack.push(ch)
        # TODO 8: 닫는 괄호를 처리하세요.
        if ch == ')' or ch == ']' or ch == '}':
            if stack.is_empty():
                return False

            top = stack.pop()
            if ch == ')' and top != '(':
                return False
            if ch == ']' and top != '[':
                return False
            if ch == '}' and top != '{':
                return False

    # TODO 9: 최종 스택 상태로 결과를 반환하세요.
    return stack.is_empty()

# 기대 결과
assert is_balanced('(a + b) * [c - d]') is True
assert is_balanced('([)]') is False
assert is_balanced('(()') is False
assert is_balanced('') is True
print('괄호 검사 테스트 통과!')
```

### 도전 2 — 후위 표기식(Postfix) 계산
공백으로 분리된 토큰을 입력받아 계산하세요. 예: `"2 3 + 4 *"` → `(2+3)*4` → `20`

규칙: 숫자는 push, 연산자는 pop 두 번 후 계산 결과를 push합니다. 두 번째로 pop한 값이 **왼쪽 피연산자**입니다.

```python
def evaluate_postfix(expression):
    stack = Stack()
    operators = {'+', '-', '*', '/'}

    for token in expression.split():
        # TODO 10: 숫자와 연산자를 구분하여 계산하세요.
        # 필요하면 float(token)을 사용하세요.
        if token in operators:
            b = stack.pop()
            a = stack.pop()

            if token == '+':
                result = a + b
            elif token == '-':
                result = a - b
            elif token == '*':
                result = a * b
            elif token == '/':
                result = a / b

            stack.push(result)
        else:
            stack.push(float(token))

    # TODO 11: 올바른 식인지 확인하고 최종값을 반환하세요.
    if stack.size() != 1:
        raise ValueError("Invalid expression")

    return stack.pop()

assert evaluate_postfix('2 3 + 4 *') == 20
assert evaluate_postfix('10 2 / 3 -') == 2
print('후위 표기식 테스트 통과!')
```

## 2. Queue ADT — FIFO
**First-In, First-Out:** 먼저 넣은 항목이 먼저 나온다.

큐의 앞(front)에서 제거하고, 뒤(rear)에 삽입합니다. Python 리스트의 `pop(0)`은 앞 원소를 꺼낸 뒤 나머지를 이동시켜 **O(n)**입니다. 실무에서는 `deque.popleft()`를 사용합니다.

```python
class Queue:
    def __init__(self):
        # TODO 12: deque를 내부 저장소로 만드세요.
        self.items = deque()

    def enqueue(self, item):
        # TODO 13: rear에 item을 추가하세요.
        self.items.append(item)

    def dequeue(self):
        # TODO 14: 비어 있으면 IndexError, 아니면 front 항목을 반환·제거하세요.
        if self.is_empty():
          raise IndexError("dequeue from empty queue")
        return self.items.popleft()

    def front(self):
        # TODO 15: front 항목을 제거하지 않고 반환하세요.
        if self.is_empty():
          raise IndexError("front from empty queue")
        return self.items[0]

    def is_empty(self):
        # TODO 16
        return len(self.items) == 0

    def size(self):
        # TODO 17
        return len(self.items)
```

```python
# Queue 기본 테스트
q = Queue()
assert q.is_empty() is True
q.enqueue('A'); q.enqueue('B'); q.enqueue('C')
assert q.front() == 'A'
assert q.dequeue() == 'A'
assert q.dequeue() == 'B'
assert q.size() == 1
assert q.dequeue() == 'C'
expect_error(IndexError, q.front)
print('Queue 기본 테스트 통과!')
```

## 3. 응용 — 작업 대기열
아래 `process_jobs`는 들어온 작업을 순서대로 처리한 결과를 반환해야 합니다.

```python
def process_jobs(jobs):
    queue = Queue()
    # TODO 18: jobs를 큐에 넣고, 큐가 빌 때까지 dequeue하여 결과 리스트에 담으세요.
    result = []

    for job in jobs:
      queue.enqueue(job)

    while not queue.is_empty():
      result.append(queue.dequeue())

    return result

assert process_jobs(['print-A', 'print-B', 'print-C']) == ['print-A', 'print-B', 'print-C']
print('작업 대기열 테스트 통과!')
```
